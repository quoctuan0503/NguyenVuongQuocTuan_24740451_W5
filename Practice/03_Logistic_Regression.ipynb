# Bài tập Logistic Regression

File nằm ở data/Social_Network_Ads.csv.

Mỗi dòng là một khách. Age và EstimatedSalary là feature. Purchased là label: 0 là không mua, 1 là mua. 



## Step 1. Load the data


```python
import pandas as pd
data = pd.read_csv('../data/Social_Network_Ads.csv')
print(data.head())
```

## Step 2. Split features, label, then train and test sets

Hint:

- features are the two columns Age and EstimatedSalary
- label is Purchased
- use train_test_split with test_size = 0.2 and random_state = 0

80% of the rows are for training. 20% are for testing later.


```python
from sklearn.model_selection import train_test_split
features = data[["Age", "EstimatedSalary"]]
label = data["Purchased"]
x_train, x_test, y_train, y_test = train_test_split(features, label, test_size=0.2, random_state=0)
```

## Step 3. Train the model

Hint: LogisticRegression(max_iter = 100), then fit on x_train and y_train.

fit loops inside the library, at most 100 rounds, and it can stop early. Print model.n_iter_[0] to see how many rounds it ran.

Print three numbers:

- intercept_ is the bias
- coef_[0][0] is the weight of Age
- coef_[0][1] is the weight of EstimatedSalary

z = intercept_ + weight_age * Age + weight_salary * EstimatedSalary


```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=100)
model.fit(x_train, y_train)
print("Số vòng lặp đã chạy (n_iter_):", model.n_iter_[0])
print("Intercept (bias):", model.intercept_[0])
print("Weight của Age:", model.coef_[0][0])
print("Weight của EstimatedSalary:", model.coef_[0][1])
```

## Step 4. Predict one customer

Hint: the customer has Age = 48 and EstimatedSalary = 50000. Use predict_proba and take element [0][1]. That value is the buy probability.

Compare it with the threshold 0.5. If it is greater than or equal to 0.5, the prediction is Buy. Otherwise it is Not buy.

Also print model.score(x_test, y_test) to see the accuracy on the test set. A reasonable result is about 0.9.


```python
customer = pd.DataFrame([[48, 50000]], columns=["Age", "EstimatedSalary"])
probability = model.predict_proba(customer)[0][1]
prediction = "Buy" if probability >= 0.5 else "Not buy"
accuracy = model.score(x_test, y_test)

print(f"Buy probability: {probability:.4f}")
print("Prediction:", prediction)
print(f"Test Accuracy: {accuracy:.4f}")
```
